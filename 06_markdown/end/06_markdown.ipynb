## Tables

**Syntax:**
- Columns are separated by pipes (`|`)
- The separator line uses hyphens (`---`) to define alignment
- Cells are filled in row by row

**Example:**
| Header 1 | Header 2 |
|----------|----------|
| Cell 1   | Cell 2   |
| Cell 3   | Cell 4   |

**Alignment options:**
| Left-aligned | Center-aligned | Right-aligned |
|:-------------|:--------------:|--------------:|
| This text    | This text      | This text     |
| aligns left  | aligns center  | aligns right  |
