## List methods

`loads = [10, 30, 50, 75, 30]`


| Method | Description | Example |
|--------|-------------|---------|
| `list.append(elem)` | Add a single element to the end of `list`. | `loads.append(100)` → `[10, 30, 50, 75, 30, 100]` adds 100 to the end |
| `list.insert(index, elem)` | Insert an element at the given index. | `loads.insert(2, 40)` → `[10, 30, 40, 50, 75, 30]` inserts 40 at position 2 |
| `list.remove(elem)` | Remove the first occurrence of the element. | `loads.remove(30)` → `[10, 50, 75, 30]` removes first `30` from list |
| `list.pop(index)` | Remove and return the element at the given index. | `popped = loads.pop(2)` → `50` removes and returns element at index 2 |
| `list.extend(list2)` | Add all elements from `list2` to the end of `list`. | `loads.extend([90, 110, 120])` → `[10, 30, 50, 75, 30, 90, 110, 120]` |
| `list.index(elem)` | Return the index of the first occurrence of the element. | `idx = loads.index(75)` → `3` returns index of first 75 value |
| `list.sort()` | Sort `list` in ascending order. | `loads.sort()` → `[10, 30, 30, 50, 75]` sorts numerically |
| `list.reverse()` | Reverse the order of `list`. | `loads.reverse()` → `[30, 75, 50, 30, 10]` reverses element order |

## Dictionary methods

`W12x50 = {
    "h": 12.19, 
    "b": 8.08,  
    "weight": 50,
    "Ix": 394
}`

| Method | Description | Example |
|--------|-------------|---------|
| `dict.keys()` | Return all keys as a view object | `W12x50.keys()` → `dict_keys(['h', 'b', 'weight', 'Ix'])` |
| `dict.values()` | Return all values as a view object | `W12x50.values()` → `dict_values([12.19, 8.08, 50, 394])` |
| `dict.items()` | Return all key-value pairs as tuples | `W12x50.items()` → `dict_items([('h', 12.19), ('b', 8.08), ('weight', 50), ('Ix', 394)])` |
| `dict.get(key, default)` | Safely get value; return default if key missing | `W12x50.get('tw', 0.415)` → `0.415` |
| `dict.update(dict2)` | Add/update multiple key-value pairs from another dict | `W12x50.update({'tw': 0.415, 'tf': 0.585})` |
| `dict.pop(key)` | Remove and return value for given key | `W12x50.pop('weight')` → `50` |
| `dict.clear()` | Remove all items from dictionary | `W12x50.clear()` → `{}` |
| `dict.copy()` | Create a shallow copy of the dictionary | `new_beam = W12x50.copy()` |

# Lists vs dictionaries

Use LISTS when:
* When position matters and you access by position (nodal coordinates)
* You need sequential processing (deflections along the beam)
* Order matters (load steps)

Use DICTIONARIES when:
* You access by name (`materials['S355']`, `sections['W12x50']`)
* Clarity is crucial
* You have a lookup table/database

```python
# When position matters - nodal coordinates
node_coords = [(0,0), (5,0), (5,3)]
node_2_position = node_coords[1]  # (5,0)
node_2_position
```

```python
# Sequential results - deflections along a beam
deflections = [0.0, 2.3, 4.1, 3.8, 0.0]  # mm
node_3_deflection = deflections[2]  # 4.1 mm
node_3_deflection
```

```python
# Ordered collections - load steps in an analysis
load_steps = [0, 10, 20, 30, 40]  # kN
step_5_load = load_steps[4]  # 40 kN
step_5_load
```
