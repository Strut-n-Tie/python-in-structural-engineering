## Formulae
The **cross-sectional area** is calculated as:
$$A = 2(B \times t_f) + (H - 2t_f) \times t_w$$

**Second moment of area** is calculated as:
$$I = \frac{(H-2t_f)^3 t_w}{12} + 2 \cdot \frac{B t_f^3}{12} + 2 B t_f \left(\frac{H}{2} - \frac{t_f}{2}\right)^2$$

where:
* $H$ = Height of the beam
* $B$ = Width of the beam
* $t_f$ = Thickness of the flange
* $t_w$ = Thickness of the web

> **Note:** Fillet radii are ignored in this calculation.

```python
# I-beam dimensions (equivalent to IPE 360)
H = 360 # overall depth [mm]
B = 170 # flange width [mm]
tf = 12.7 # flange thickness [mm]
tw = 8  # web thickness [mm]
```

```python
# Cross-sectional area
for i in range(50000000):
    A = tf*B*2 + (H-2*tf)*tw
print(f"IPE 360; area A = {A/1E2} mm^2")
```

```python
# Second moment of area
I = (H-2*tf)**3*tw/12 + 2*B*tf**3/12 + 2*B*tf*(H/2-tf/2)**2
print(f"IPE 360; I = {I/1E4} cm^4")
```

```python
# Section modulus
W = 2*I/H
print(f"IPE 360; W = {W/1E3} cm^3")
```
