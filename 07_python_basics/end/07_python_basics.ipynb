# Variables

Variables store information in memory. 

A variable: 
* has a **name**
* stores a **value** of a particular data type

A variable is created by assigning a value to a name using the `=` operator:

`name = value`

When naming variables:
- Use only letters, numbers, and underscores (`_`)
- Start the name with a letter or underscore, not a number
- Do not use spaces
- Use descriptive names whenever possible
- Python is case-sensitive (`beam_length` vs `Beam_Length`)

````python
# VARIABLES

beam_length = 10
print(beam_length)

# Assigning multiple variables in a single line:
beam_length, E, I = 10, 210e9, 0.025

beam_length = 8
print(beam_length)
````

# Printing

The `print()` function displays information to the screen.

**Basic printing**

```python
print(beam_length)
print("Length = ", beam_length)
```

**Formatted strings (f-strings)**

F-strings allow variables to be inserted directly into text.

```python
print(f"Beam length = {beam_length} m")
```

**Number formatting**

The general format is:

```python
{value:.precision[type]}
```

where:
- `precision` is the number of digits after the decimal point
- `type` is an optional letter specifying the display format:
    - `f` displays the value in fixed-point notation (e.g., `3.14`)
    - `e` displays the value in scientific notation (e.g., `3.14e+00`)
    - `g` general format (chooses `f` or `e` automatically)

**Examples:**

```python
f"{3.14159:.2f}"  # 3.14
f"{210E9:.2e}"    # 2.10e+11

````python
# PRINTING

beam_length = 8 # m
E = 210e9 # Pa
I = 0.025 # m^4

# Basic printing
print(beam_length)

# Printing multiple values
print("Length =", beam_length, "m")

# f-string formatting
print(f"Beam length = {beam_length} m")
print(f"E = {E:.2e} Pa")
print(f"I = {I:.3f} m^4")

# Combining multiple variables
print(f"Beam length = {beam_length} m, E = {E:.2e} Pa, I = {I:.3f} m^4")
````

# Basic data types

| Type | Description | Example |
|------|-------------|----------|
| `int` | Integer numbers | `10` |
| `float` | Decimal numbers | `10.5` |
| `str` | Text (string) | `"UB 457 x 191 x 98"` |
| `bool` | True/False values | `True` |

````python
# BASIC DATA TYPES

number_of_beams = 6 # int
beam_width = 0.3 # float 
section = "UB 457 x 191 x 98" # str 
is_steel = True # bool

print(f"number_of_beams = {number_of_beams} ({type(number_of_beams).__name__})")
print(f"beam_width = {beam_width} ({type(beam_width).__name__})")
print(f"section = {section} ({type(section).__name__})")
print(f"is_steel = {is_steel} ({type(is_steel).__name__})")
````

# Example - RC Section Design

<img src="img/beam_UDL.svg" width="300">

**Input Parameters**

| Parameter | Value | Units |
|-----------|-------|-------|
| $L$ | 8 | m |
| $d$ | 550 | mm |
| $b$ | 450 | mm |
| $w_{Ed}$ | 50 | kN/m |
| $f_{ck}$ | 30 | MPa |
| $f_{yk}$ | 500 | MPa |
| $\gamma_s$ | 1.15 | - |

**Rectangular stress block:**

<img src="img/stress_distribution.png" width="500">

**Bending moment:**

$$M_{Ed} = \frac{w_{Ed} \cdot L^2}{8} $$

**Design yield strength of reinforcement:**

$$f_{yd} = \frac{f_{yk}}{\gamma_s} $$

**Non-dimensionalised moment:**

$$k = \frac{M_{Ed}}{b \cdot d^2 \cdot f_{ck}} \le k' = 0.167$$

**Flexural Lever Arm:**

$$z = 0.5 \cdot d \left(1 + \sqrt{1 - 3.53k} \right)$$

**Required Reinforcement Area:**

$$A_{s,req} = \frac{M_{Ed}}{f_{yd} \cdot z}$$

**Provided Reinforcement Area:**

$$A_{s,prov} = \frac{n \cdot \pi \cdot \phi^2}{4}$$


````python
# EXAMPLE

# Input
L = 8 # m
d = 550 # mm
b = 450 # mm
w_Ed = 50 # kN/m
f_ck = 30 # MPa
f_yk = 500 # MPa
gamma_s = 1.15

# Solution
import math

# Bending moment
M_Ed = (w_Ed * L**2) / 8 * 1E6 # convert kNm to Nmm
print(f"M_Ed = {M_Ed/1E6:.1f} kNm")

# Design yield strength
f_yd = f_yk / gamma_s 
print(f"f_yd = {f_yd:.1f} MPa")

# Non-dimensionalised moment
k = M_Ed/(b*d**2*f_ck) 
print(f"k = {k:.4f}")

# Flexural lever arm
z = 0.5*d*(1+(1-3.53*k)**0.5) 
print(f"z = {z:.0f} mm")

# Required reinforcement area
A_s_req = M_Ed/(f_yd * z) 
print(f"A_s_req = {A_s_req:.0f} mm^2")

# Provided reinforcement area
A_s_prov = 4 * 25**2 * math.pi / 4 # Try 4B25
print(f"A_s_prov = {A_s_prov:.0f} mm^2")
````
